```python
import torch
import numpy as np
```

### 1. Importing Libraries 📚

In this step, we import the fundamental libraries for our workflow:
* **PyTorch (`torch`)**: A powerful open-source machine learning library primarily used for deep learning applications.
* **NumPy (`numpy`)**: The fundamental package for scientific computing with Python.

```python
X = torch.tensor([[1.0, 4.0, 7.0], [2.0, 3.0, 6.0]])
```

### 2. Creating a PyTorch Tensor 🛠️

We initialize a $2 \times 3$ (2 rows, 3 columns) PyTorch tensor named `X` containing floating-point numbers.

```python
X
```

### 3. Inspecting Tensor Shape (`shape`) 📐

The `shape` attribute (or `size()` method) reveals the dimensionality of the tensor. Our output `torch.Size([2, 3])` confirms that `X` consists of **2 rows** and **3 columns**.

```python
X.shape
```

### 4. Inspecting Data Type (`dtype`) 💾

This attribute specifies the data type of the elements stored inside the tensor. By default, PyTorch initializes floating-point numbers as `torch.float32` (single-precision float) to optimize speed and memory usage.

```python
X.dtype
```

### 5. Tensor Slicing 🔪

Slicing allows us to extract specific subsets of data from our tensor:
* `X[:, :2]` extracts **all rows** (`:`) and **the first 2 columns** (`:2` corresponding to indices 0 and 1).

```python
X[:,:2]
```

### 6. Finding Maximum Values Along an Axis (`max`) 🔍

The `.max(dim=1)` method calculates the maximum value across the column dimension (for each row individually):
* **`values`**: Returns the maximum element found in each row (`[7.0, 6.0]`).
* **`indices`**: Returns the column indices where these maximum values are located (`[2, 2]`).

```python
X.max(dim=1)
```

### 7. Transposing Tensors (`.T`) 🔄

Transposing swaps the rows and columns of a matrix. Our $2 \times 3$ tensor is transformed into a $3 \times 2$ matrix.

```python
X.T
```

```python
X
```

```python
print(f"Original shape: {X.shape}")
print(f"Transposed shape: {X.T.shape}")
```

```python
a = np.array([[1., 4., 7.], [2., 3., 6]])
```

### 8. NumPy vs. PyTorch Default Data Types ⚙️

Notice the default data type difference:
* **NumPy** arrays default to `float64` (double-precision float) on most platforms.
* **PyTorch** defaults to `float32` (single-precision float) to maximize hardware performance and memory efficiency during neural network training.

```python
a.dtype
```

```python
# Converting the NumPy array to a PyTorch tensor to see type conversion
torch_from_np = torch.from_numpy(a)
print("Converted PyTorch Tensor:", torch_from_np)
print("Resulting data type:", torch_from_np.dtype)
```

### 9. Memory Sharing Between NumPy and PyTorch 🧠🔄

One of PyTorch's most powerful features is **zero-copy memory sharing** when converting data back and forth with NumPy.

Let's re-initialize our NumPy array `a` and create a PyTorch tensor `X` from it using `torch.from_numpy()`.

```python
a = np.array([[1., 4., 7.],[2., 3., 6.]])
```

```python
X = torch.from_numpy(a)
```

Let's check our newly created PyTorch tensor `X`:

```python
X
```

### 10. Modifying the Source Array ⚡

Because `torch.from_numpy()` shares the same memory space, any in-place modification made to the NumPy array `a` will **automatically propagate** to the PyTorch tensor `X` without copying any data.

Let's change the element at index `[1, 2]` in the array `a` from `6.0` to `5.0`:

```python
a[1,2] = 5.
```

Now let's check the PyTorch tensor `X` again to verify if the value updated automatically:

```python
X
```
