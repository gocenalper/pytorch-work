## 🛠️ Device Management and Tensor Definition in PyTorch

In this section, we utilize the PyTorch library to detect available hardware accelerators (such as GPU or Apple Silicon MPS) so that our code runs efficiently, and we transfer our tensors to these devices.

### 📱 1. Hardware Accelerator Detection
To ensure hardware-agnostic execution, we first select the most powerful computing unit available in the environment:
* **CUDA (GPU):** For parallel computing on NVIDIA graphics cards.
* **MPS (Metal Performance Shaders):** For Apple Silicon (M1/M2/M3) Mac computers.
* **CPU:** The default processor when no accelerator is detected.

### 📦 2. Creating Tensors and Device Transfer
We initialize a matrix (`M`) using `torch.tensor()` and then use the `.to(device)` method to transfer it to the memory of the chosen computing device. This step is crucial for accelerating deep learning model training.

```python
import torch
```

### 📱 1. Hardware Accelerator Detection
We check for available hardware accelerators (CUDA or MPS) so that our computations are offloaded to the GPU when possible, falling back to CPU otherwise.

```python
if torch.cuda.is_available():
  device = 'cuda'
elif torch.backends.mps.is_available():
  device = 'mps'
else:
  device = 'cpu'
print(device)
```

```python
device
```

### 📦 2. Creating a PyTorch Tensor
Here, we initialize a 2D tensor (matrix) `M` with floating-point numbers on the default device (usually the CPU).

```python
M = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
```

### 🚀 3. Transferring the Tensor to the Target Device
Using the `.to(device)` method, we copy the tensor over to the hardware accelerator (GPU/MPS) that we detected earlier to prepare it for high-performance operations.

```python
M = M.to(device)
```
